```python
import os

def check_paths_in_df(df, column_name, db_root):
    print(f"--- '{column_name}' 컬럼 파일 존재 여부 확인 ---")
    if column_name not in df.columns:
        print(f"Error: Column '{column_name}' not found in DataFrame.")
        return

    unique_paths = df[column_name].dropna().unique()

    if len(unique_paths) == 0:
        print(f"No valid paths found in '{column_name}' to check.")
        return

    # Define the problematic external root and its Colab equivalent
    original_external_root = '/media/jisha/E/trash_dataset/'
    colab_equivalent_prefix = 'drive/MyDrive/'

    all_exist = True
    for path_suffix in unique_paths:
        # Step 1: Replace the old problematic absolute prefix with the new Colab-equivalent one
        if path_suffix.startswith(original_external_root):
            # Replace the external root part. Example: from /media/.../dataset_name/frames to drive/MyDrive/dataset_name/frames
            temp_path_colab_root = path_suffix.replace(original_external_root, colab_equivalent_prefix)
        else:
            # If it doesn't start with the problematic prefix, assume it's already in a potentially correct format
            temp_path_colab_root = path_suffix

        # Step 2: Now, temp_path_colab_root should be like 'drive/MyDrive/camera1_plastics_shareable_dataset/frames/...' or 'frames/...' if already relative.
        # We need to ensure the path is correctly joined with the db_root, which is 'drive/MyDrive/camera1_plastics_shareable_dataset'.
        # The `db_root` variable itself is a relative path to /content/.

        full_path_to_check = None
        # If temp_path_colab_root already contains the db_root structure, extract the relative part
        if temp_path_colab_root.startswith(db_root):
            # e.g., temp_path_colab_root = 'drive/MyDrive/camera1_plastics_shareable_dataset/frames/...'
            # and db_root = 'drive/MyDrive/camera1_plastics_shareable_dataset'
            # We want to check 'db_root/frames/...' so we extract 'frames/...' from temp_path_colab_root
            relative_part_from_db_root = temp_path_colab_root[len(db_root):]
            # Remove leading slash if present for os.path.join
            if relative_part_from_db_root.startswith('/') or relative_part_from_db_root.startswith('\\'):
                relative_part_from_db_root = relative_part_from_db_root[1:]
            full_path_to_check = os.path.join(db_root, relative_part_from_db_root)
        else:
            # If temp_path_colab_root doesn't start with db_root, it might already be the exact path relative to /content/
            # or a path that needs to be directly joined with db_root (e.g. 'frames/GX010023/...' directly)
            full_path_to_check = os.path.join(db_root, temp_path_colab_root)

        # Ensure we are checking a non-empty path
        if not full_path_to_check:
            print(f"Warning: Could not construct a valid path for '{path_suffix}'. Skipping check.")
            all_exist = False
            continue

        if not os.path.exists(full_path_to_check):
            print(f"Warning: File does NOT exist: {full_path_to_check}")
            all_exist = False

    if all_exist:
        print(f"All {len(unique_paths)} unique files in '{column_name}' exist in '{db_root}'.")
    else:
        print(f"Some files in '{column_name}' do NOT exist in '{db_root}'. Please check warnings above.")
    print("\n")

# Check paths for each specified column
check_paths_in_df(meta_df, 'copied_frame_path', db_root)
check_paths_in_df(meta_df, 'copied_mask_path', db_root)
check_paths_in_df(meta_df, 'timestamp_clip_path', db_root)
```

## Extracting Audio from Video Clips

Now, I will extract the audio from the video files listed in the `timestamp_clip_path` column. The extracted audio files will be saved in the `/content/audio` directory. I'll use `moviepy` for this task.

```python
# Install moviepy if not already installed
!pip install moviepy

import os
from moviepy.editor import VideoFileClip

# Create the directory for saving audio files if it doesn't exist
audio_output_dir = '/content/audio'
os.makedirs(audio_output_dir, exist_ok=True)
print(f"Audio output directory '{audio_output_dir}' ensured.")
```

```python
extracted_audio_info = []

# Set test_limit to an integer (e.g., 5) to process only the first N files.
# Set test_limit = None to process all files in meta_df.
test_limit = None # Change to None to process all files

test_limit = 5

if test_limit is not None:
    print(f"Processing only the first {test_limit} video clips for audio extraction...")
else:
    print("Processing all video clips for audio extraction...")

# Import tqdm for progress bar
from tqdm.notebook import tqdm

# Iterate through each row of the meta_df DataFrame with tqdm
# For rows where 'timestamp_clip_path' is NaN, we pass None for clip_path
for index, row in tqdm(meta_df.iterrows(), total=len(meta_df) if test_limit is None else test_limit):
    if test_limit is not None and index >= test_limit:
        print(f"Stopped after processing {test_limit} video clips as per test_limit.")
        break

    clip_file_suffix = row['timestamp_clip_path']

    # Only process if clip_file_suffix is not NaN
    if pd.notna(clip_file_suffix):
        # Construct full video path using the established remapping logic
        original_external_root = '/media/jisha/E/trash_dataset/'
        colab_equivalent_prefix = 'drive/MyDrive/'

        if clip_file_suffix.startswith(original_external_root):
            clip_path_temp = clip_file_suffix.replace(original_external_root, colab_equivalent_prefix)
        else:
            clip_path_temp = clip_file_suffix

        if clip_path_temp.startswith(db_root):
            relative_part_clip = clip_path_temp[len(db_root):]
            if relative_part_clip.startswith('/') or relative_part_clip.startswith('\\'):
                relative_part_clip = relative_part_clip[1:]
            full_video_path = os.path.join(db_root, relative_part_clip)
        else:
            full_video_path = os.path.join(db_root, clip_path_temp)

        # Construct output audio file path
        video_basename = os.path.basename(full_video_path)
        audio_filename = os.path.splitext(video_basename)[0] + '.mp3' # or .wav
        output_audio_path = os.path.join(audio_output_dir, audio_filename)

        try:
            if os.path.exists(full_video_path):
                # print(f"Extracting audio from: {full_video_path}") # Suppress this for cleaner tqdm output
                video = VideoFileClip(full_video_path)
                if video.audio is not None:
                    video.audio.write_audiofile(output_audio_path, logger=None) # logger=None to suppress moviepy verbose output
                    extracted_audio_info.append({
                        'original_video_path': full_video_path,
                        'extracted_audio_path': output_audio_path,
                        'status': 'success'
                    })
                    # print(f"Audio saved to: {output_audio_path}") # Suppress this for cleaner tqdm output
                else:
                    extracted_audio_info.append({
                        'original_video_path': full_video_path,
                        'extracted_audio_path': None,
                        'status': 'no_audio_track'
                    })
                    print(f"Warning: No audio track found in {full_video_path}")
                video.close() # Close the video file
            else:
                extracted_audio_info.append({
                    'original_video_path': full_video_path,
                    'extracted_audio_path': None,
                    'status': 'video_not_found'
                })
                print(f"Error: Video file not found: {full_video_path}")

        except Exception as e:
            extracted_audio_info.append({
                'original_video_path': full_video_path,
                'extracted_audio_path': None,
                'status': f'error: {e}'
            })
            print(f"Error extracting audio from {full_video_path}: {e}")
    else:
        # Handle cases where 'timestamp_clip_path' is NaN
        extracted_audio_info.append({
            'original_video_path': None,
            'extracted_audio_path': None,
            'status': 'no_clip_path'
        })

print(f"\nTotal video clips processed for audio extraction: {len(extracted_audio_info)}")
# To see the results, uncomment the line below:
# display(pd.DataFrame(extracted_audio_info).head())

!mv /content/audio /content/drive/MyDrive/csire_code/audio
```

### Audio Transcription using Whisper

First, I will install the `openai-whisper` library, which is a state-of-the-art automatic speech recognition (ASR) system. It also requires `torch` for GPU acceleration and `ffmpeg` for audio processing (which should already be available from `moviepy` installation).

```python
# Install Whisper and other necessary libraries if they aren't already
!pip install -qqq openai-whisper

import os
import json
import pandas as pd
import whisper
from tqdm.notebook import tqdm
from datetime import datetime

print("Whisper and other libraries imported.")
```

Next, I will define the input and output directories and ensure the output directory for transcriptions exists.

```python
audio_input_dir = '/content/drive/MyDrive/csire_code/audio'
transcribe_output_dir = '/content/drive/MyDrive/csire_code/transcribe/v1'
manifest_output_path = '/content/drive/MyDrive/csire_code/transcribe/manifest_v1.json'

os.makedirs(transcribe_output_dir, exist_ok=True)
os.makedirs(os.path.dirname(manifest_output_path), exist_ok=True)

print(f"Audio input directory: {audio_input_dir}")
print(f"Transcription output directory: {transcribe_output_dir}")
print(f"Manifest output path: {manifest_output_path}")
```

Now, I will load the Whisper model (using the 'base' model as a default, but you can change this to 'small', 'medium', or 'large' depending on your needs and available resources). Then, I'll iterate through all audio files in the input directory, transcribe them, and save the results as text files. A manifest file will also be created to log the transcription details.

```python
# Load the Whisper model
# You can choose 'tiny', 'base', 'small', 'medium', 'large'
# Larger models offer better accuracy but require more VRAM and take longer.
print("Loading Whisper model...")
model = whisper.load_model("base")
print("Whisper model loaded.")

transcription_results = []

# Get list of audio files
audio_files = [f for f in os.listdir(audio_input_dir) if f.endswith('.mp3')]

print(f"Found {len(audio_files)} audio files to transcribe.")

for audio_filename in tqdm(audio_files, desc="Transcribing audio files"):
    audio_file_path = os.path.join(audio_input_dir, audio_filename)
    output_txt_filename = os.path.splitext(audio_filename)[0] + '.txt'
    output_txt_path = os.path.join(transcribe_output_dir, output_txt_filename)

    try:
        # Transcribe the audio
        result = model.transcribe(audio_file_path)
        transcription_text = result["text"]

        # Save the transcription to a text file
        with open(output_txt_path, 'w', encoding='utf-8') as f:
            f.write(transcription_text)

        transcription_results.append({
            'audio_filename': audio_filename,
            'audio_path': audio_file_path,
            'transcription_path': output_txt_path,
            'status': 'success',
            'transcription_text_snippet': transcription_text[:100] + '...' if len(transcription_text) > 100 else transcription_text
        })
    except Exception as e:
        transcription_results.append({
            'audio_filename': audio_filename,
            'audio_path': audio_file_path,
            'transcription_path': None,
            'status': f'error: {e}',
            'transcription_text_snippet': None
        })
        print(f"Error transcribing {audio_filename}: {e}")

print(f"\nCompleted transcription for {len(transcription_results)} files.")
```

Finally, I will create a manifest file `manifest_v1.json` containing the details of this transcription run, including the model used, timestamp, and a summary of each transcribed file.

```python
# Create the manifest file
manifest_data = {
    'run_timestamp': datetime.now().isoformat(),
    'whisper_model': 'base', # Ensure this matches the model used above
    'audio_input_directory': audio_input_dir,
    'transcription_output_directory': transcribe_output_dir,
    'transcription_summary': transcription_results
}

with open(manifest_output_path, 'w', encoding='utf-8') as f:
    json.dump(manifest_data, f, ensure_ascii=False, indent=4)

print(f"Manifest file saved to: {manifest_output_path}")

# Display a snippet of the manifest for verification
print("\nManifest file content (first 5 entries):")
with open(manifest_output_path, 'r', encoding='utf-8') as f:
    manifest_content = json.load(f)
display(pd.DataFrame(manifest_content['transcription_summary']).head())
```

`timestamp_groups_df`의 `clip_path`와 `meta_df`의 `timestamp_clip_path`가 유사한 형식임을 알 수 있습니다. 이들을 매핑하여 전사 결과를 그룹화된 타임스탬프 정보와 연결할 수 있습니다. 예를 들어, 특정 비디오 클립에 대한 전사 내용과 해당 클립의 `phrase`나 `reason` 등을 비교할 수 있습니다.

```python
import pandas as pd # Ensure pandas is imported if not already in scope
import os # Ensure os is imported if not already in scope

# Ensure meta_df is loaded. This is necessary if previous cells defining it were not run.
csv_file_path = f'{db_root}/camera1_plastics_meta2.csv'
try:
    meta_df = pd.read_csv(csv_file_path)
except FileNotFoundError:
    print(f"Error: The file '{csv_file_path}' for meta_df was not found.")
    meta_df = pd.DataFrame() # Create empty DataFrame to avoid further errors
except Exception as e:
    print(f"An unexpected error occurred while loading meta_df: {e}")
    meta_df = pd.DataFrame() # Create empty DataFrame to avoid further errors

# Ensure timestamp_groups_df is loaded. This is necessary if previous cells defining it were not run.
timestamp_groups_csv_path = f'{db_root}/camera1_plastics_timestamp_groups.csv'
try:
    timestamp_groups_df = pd.read_csv(timestamp_groups_csv_path)
except FileNotFoundError:
    print(f"Error: The file '{timestamp_groups_csv_path}' for timestamp_groups_df was not found.")
    timestamp_groups_df = pd.DataFrame() # Create empty DataFrame to avoid further errors
except Exception as e:
    print(f"An unexpected error occurred while loading timestamp_groups_df: {e}")
    timestamp_groups_df = pd.DataFrame() # Create empty DataFrame to avoid further errors


# manifest_v1.json 파일을 다시 로드하는 대신, 이미 메모리에 있는 transcription_results 변수를 사용합니다.
transcription_df = pd.DataFrame(transcription_results)

# 'audio_filename'에서 clip ID를 추출 (예: 'GX010023__000012.720__000014.120.mp3' -> 'GX010023__000012.720__000014.120')
transcription_df['clip_id'] = transcription_df['audio_filename'].apply(lambda x: os.path.splitext(x)[0])

# meta_df의 'timestamp_clip_path'에서 clip ID 추출
# 기존의 remapping 로직을 사용하여 'timestamp_clip_path'를 정규화된 형태로 만든 후 clip ID를 추출합니다.
original_external_root = '/media/jisha/E/trash_dataset/'
colab_equivalent_prefix = 'drive/MyDrive/'

def get_normalized_clip_id(clip_path_suffix):
    if pd.isna(clip_path_suffix): return None

    # Step 1: Replace the old problematic absolute prefix
    if clip_path_suffix.startswith(original_external_root):
        clip_path_temp = clip_path_suffix.replace(original_external_root, colab_equivalent_prefix)
    else:
        clip_path_temp = clip_path_suffix

    # Step 2: Extract the relevant part after db_root (if present) or use as is
    if clip_path_temp.startswith(db_root):
        relative_part_clip = clip_path_temp[len(db_root):]
        if relative_part_clip.startswith('/') or relative_part_clip.startswith('\\'):
            relative_part_clip = relative_part_clip[1:]
        # Now, relative_part_clip should look like 'audio_video_clips/GX010023/GX010023__000012.720__000014.120.mp4'
        # We need to get just the filename part, then remove extension
        return os.path.splitext(os.path.basename(relative_part_clip))[0]
    else:
        # If it doesn't start with db_root, assume it's already a relative path or direct filename
        return os.path.splitext(os.path.basename(clip_path_temp))[0]

meta_df['normalized_clip_id'] = meta_df['timestamp_clip_path'].apply(get_normalized_clip_id)

# timestamp_groups_df의 'clip_path'에서 clip ID 추출
# 'clip_path' 컬럼이 이미 상대 경로 또는 파일 이름 형식일 것이므로, 바로 확장자를 제거합니다.
timestamp_groups_df['clip_id'] = timestamp_groups_df['clip_path'].apply(lambda x: os.path.splitext(os.path.basename(x))[0] if pd.notna(x) else None)


# 이제 transcription_df와 timestamp_groups_df를 'clip_id'를 기준으로 병합하여 비교할 수 있습니다.
# meta_df의 정보를 포함하여 더 풍부한 비교를 할 수도 있습니다.

# transcription_df와 meta_df 병합
merged_transcriptions = pd.merge(
    transcription_df,
    meta_df[['normalized_clip_id', 'phrase', 'reason', 'category_name', 'supercategory_name']],
    left_on='clip_id',
    right_on='normalized_clip_id',
    how='left'
)

# 병합된 결과에 timestamp_groups_df 정보 추가
final_comparison_df = pd.merge(
    merged_transcriptions,
    timestamp_groups_df[['clip_id', 'meta2_phrase', 'begin_time', 'end_time']].rename(columns={'meta2_phrase': 'timestamp_group_phrase'}),
    left_on='clip_id',
    right_on='clip_id',
    how='left'
)

print("\nMerged DataFrame for comparison (first 5 rows):")
display(final_comparison_df.head())

print("\nColumns available for comparison:")
print(final_comparison_df.columns.tolist())

print("\n비교 아이디어:")
print("1. 'transcription_text_snippet'과 'phrase' 또는 'timestamp_group_phrase'를 비교하여 키워드 일치 여부 확인")
print("2. 'category_name', 'supercategory_name'과 'transcription_text_snippet'의 관련성 분석")
print("3. 특정 키워드가 포함된 전사 내용을 기준으로 데이터 필터링")
```

### 전사 결과와 기존 메타데이터의 'phrase' 비교

이제 `final_comparison_df`를 활용하여 Whisper 전사 (`transcription_text_snippet`) 내용과 `meta_df`의 `phrase` 및 `timestamp_groups_df`의 `timestamp_group_phrase` 간의 일치 여부를 비교해 보겠습니다. 주로 키워드 포함 여부를 확인하는 방식으로 진행합니다.

```python
# 텍스트 스니펫과 구문 간의 키워드 일치 여부를 확인하는 함수
def check_keyword_match(transcription_text, target_phrase):
    if pd.isna(transcription_text) or pd.isna(target_phrase): return False
    # 소문자로 변환하여 비교하고, 띄어쓰기를 기준으로 단어 분리 (간단한 형태)
    transcription_words = set(transcription_text.lower().split())
    target_words = set(target_phrase.lower().split())

    # 전사된 텍스트의 단어가 target_phrase의 단어에 하나라도 포함되면 True
    return bool(transcription_words.intersection(target_words))

# 'transcription_text_snippet'과 'phrase' 간의 일치 여부 확인
final_comparison_df['matches_meta_phrase'] = final_comparison_df.apply(
    lambda row: check_keyword_match(row['transcription_text_snippet'], row['phrase']),
    axis=1
)

# 'transcription_text_snippet'과 'timestamp_group_phrase' 간의 일치 여부 확인
final_comparison_df['matches_timestamp_group_phrase'] = final_comparison_df.apply(
    lambda row: check_keyword_match(row['transcription_text_snippet'], row['timestamp_group_phrase']),
    axis=1
)

print("키워드 매칭 결과 (상위 10개):")
display(final_comparison_df[['audio_filename', 'transcription_text_snippet', 'phrase', 'matches_meta_phrase', 'timestamp_group_phrase', 'matches_timestamp_group_phrase']].head(10))

print("\n'phrase'와 매칭된 전사 비율:")
print(f"{final_comparison_df['matches_meta_phrase'].sum() / len(final_comparison_df):.2%}")

print("\n'timestamp_group_phrase'와 매칭된 전사 비율:")
print(f"{final_comparison_df['matches_timestamp_group_phrase'].sum() / len(final_comparison_df):.2%}")
```

### Using Sentence Embeddings for Category Assignment

To improve the classification of transcriptions, we will now use a pretrained sentence embedding model to encode the `meta2_phrase` phrases and then use cosine similarity to find the most similar category for each transcription.

```python
# Install the sentence-transformers library
!pip install -qqq sentence-transformers

import pandas as pd
from sentence_transformers import SentenceTransformer, util
import numpy as np

print("Sentence-transformers and other libraries imported.")
```

First, we will load a pretrained sentence embedding model. We'll use `all-MiniLM-L6-v2` which is efficient and performs well for semantic similarity tasks.

```python
# Load a pretrained sentence embedding model
print("Loading SentenceTransformer model...")
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
print("SentenceTransformer model loaded.")

# Prepare the reference phrases and their corresponding categories from timestamp_groups_df
# Ensure timestamp_groups_df is loaded from previous cells

# Filter out NaN phrases and drop duplicates to get unique phrase-category mappings
reference_phrases_df = timestamp_groups_df[['meta2_phrase', 'category_names']].dropna().drop_duplicates().reset_index(drop=True)

# Encode all unique meta2_phrases to create reference embeddings
print(f"Encoding {len(reference_phrases_df)} unique reference phrases...")
reference_phrase_embeddings = embedding_model.encode(reference_phrases_df['meta2_phrase'].tolist(), convert_to_tensor=True)
print("Reference phrases encoded.")

# Create a dictionary for quick lookup of category names
phrase_to_category_map = reference_phrases_df.set_index('meta2_phrase')['category_names'].to_dict()

print("Reference phrases and categories prepared.")
```

Now, we define a function that takes a transcription, encodes it, compares it with the reference phrase embeddings using cosine similarity, and assigns the category of the most similar reference phrase.

```python
def assign_category_with_embedding(transcription_text):
    if pd.isna(transcription_text) or not transcription_text.strip():
        return "Unknown"

    # Encode the input transcription text
    transcription_embedding = embedding_model.encode(transcription_text, convert_to_tensor=True)

    # Compute cosine similarities between the transcription and all reference phrases
    cosine_scores = util.cos_sim(transcription_embedding, reference_phrase_embeddings)[0]

    # Get the index of the most similar reference phrase
    best_match_idx = np.argmax(cosine_scores.cpu().numpy())

    # Get the corresponding reference phrase and its category
    best_match_phrase = reference_phrases_df.loc[best_match_idx, 'meta2_phrase']
    predicted_category = phrase_to_category_map.get(best_match_phrase, "Unknown")

    return predicted_category

print("Embedding-based category assignment function defined.")
```

Next, we will apply this new `assign_category_with_embedding` function to both V1 (Base) and V2 (Large) transcription results to get embedding-based predicted categories. Then we will compare their performance with the actual `category_names`.

```python
def map_transcript_to_category(transcript_text):
    if pd.isna(transcript_text): return "Unknown"
    text = transcript_text.lower()

    # --- 1. Exact Matches (Highest Priority) ---
    # These are highly specific phrases that should map to a single, confident category.
    if text == "non -deposit garbage bag":
        return "Non-Deposit" # Fix for common misclassification (Actual: Non-Deposit, Predicted: Garbage Bags)
    if text == "dissolt.":
        return "Deposit #1 Bottles"
    if text == "fly plastic.":
        return "Black Plastic Containers"
    if text == "under posit one down to positive":
        return "Non-Deposit"
    if text == "no." or text == "no":
        return "Containers #1/Clamshells"
    if text == "one deposit." or text == "one deposit":
        return "Deposit #1 Bottles"
    if "other one" in text or "lid" in text:
        return "Containers #1/Clamshells"
    if "five" in text or "5" in text:
        return "Containers #5/PP"
    if text == "unnumbered.":
        return "Other Unnumbered Containers & Fragments"
    if text == "other plastic.":
        return "Other Plastics"
    if "two" in text and "color" in text:
        return "Containers #2 Colored"
    if text == "number one.":
        return "Deposit #1 Bottles" # Assuming "number one." alone usually means Deposit #1 Bottles based on phrase_counts_by_category

    # --- 2. Highly Specific Keyword-Based Rules (Higher Priority) ---
    # NEW RULE: If both "one" and "non" are present, classify as "Other #1 Bottles"
    if ("one" in text or "1" in text) and "non" in text:
        return "Other #1 Bottles"

    if "non" in text and "deposit" in text:
        return "Non-Deposit"

    # These cover phrases containing strong category indicators.
    if "garbage bag" in text:
        return "Garbage Bags"

    if "black" in text:
        return "Black Plastic Containers"

    if "clam" in text and "shell" in text:
        return "Containers #1/Clamshells"

    if "two natural" in text:
        return "Containers #2 Natural"
    if "two color" in text or "two colored" in text:
        return "Containers #2 Colored"

    # --- 3. Number-Based Rules (Medium Priority) ---
    if "three" in text or "3" in text or "seven" in text or "7" in text or "four" in text or "4" in text or "six" in text or "6" in text:
        return "Containers #3-7 (Not #5)"
    if "five" in text or "5" in text:
        return "Containers #5/PP"

    # --- 4. Deposit/Non-Deposit & Number #1 Rules (Careful Prioritization) ---
    # This section is critical for distinguishing similar phrases.
    if "number one" in text:
        if "deposit" in text:
            return "Deposit #1 Bottles"
        # The 'non -deposit' case here for 'number one' will be caught by the new 'one' and 'non' rule if 'non' is also present
        return "Other #1 Bottles" # If 'number one' but no deposit/non-deposit specifier (and not caught by 'one' and 'non' rule)

    if "deposit" in text:
        return "Deposit #1 Bottles"


    # --- 5. Other Generic Category Keywords (Lower Priority) ---
    if "plastic film" in text or "film plastic" in text or "film" in text:
        return "Other Plastic Film"

    if "other plastic" in text:
        return "Other Plastics"

    if "unnumbered" in text:
        return "Other Unnumbered Containers & Fragments"

    if "bulky rigid" in text:
        return "Other Rigid/Bulky Plastics"

    if "if you find numbers" in text:
        return "Other Unnumbered Containers & Fragments"

    if "plastic" in text:
        return "Plastic"

    if "one" in text:
        return "Containers #1/Clamshells"
    # Default return if no rules match
    return "Unknown"

# Apply the mapping function to timestamp_groups_df['meta2_phrase']
timestamp_groups_df['predicted_category_from_phrase'] = timestamp_groups_df['meta2_phrase'].apply(map_transcript_to_category)

# Calculate the success rate
success_rate_from_phrase = (timestamp_groups_df['predicted_category_from_phrase'] == timestamp_groups_df['category_names']).mean() * 100

print(f"Success Rate of `map_transcript_to_category` using `timestamp_groups_df['meta2_phrase']`: {success_rate_from_phrase:.2f}%")

# Identify misclassified entries
incorrect_classifications = timestamp_groups_df[
    timestamp_groups_df['predicted_category_from_phrase'] != timestamp_groups_df['category_names']
]

# Group by actual category, predicted category, and phrase to find common misclassification patterns
misclassification_patterns = incorrect_classifications.groupby(['meta2_phrase', 'category_names', 'predicted_category_from_phrase']).size().reset_index(name='count')

# Sort by count to see the most frequent misclassifications
misclassification_patterns = misclassification_patterns.sort_values(by='count', ascending=False)

print("\nMost Frequent Misclassification Patterns (Top 20):")
display(misclassification_patterns.head(30))

print(f"\nTotal misclassified entries: {len(incorrect_classifications)}")
```
