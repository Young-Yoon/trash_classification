```python
from glob import glob  # glob.glob
f = glob(f"{file_path}/*.mp4")
print(len(f))
```

```python
!pip install ffmpeg-python

import ffmpeg

def extract_audio_with_wrapper(input_video, output_audio):
    try:
        # Build the stream workflow
        stream = ffmpeg.input(input_video)

        # Audio-only extraction with explicit bitrate parameters
        stream = ffmpeg.output(stream, output_audio, vn=None, **{'b:a': '192k'})

        # Run the conversion
        ffmpeg.run(stream, overwrite_output=True)
        print(f"Success! Audio saved to {output_audio}")
    except ffmpeg.Error as e:
        print(f"FFmpeg Error: {e.stderr.decode()}")

# Example usage
#extract_audio_with_wrapper('input_video.mp4', 'output_audio.mp3')
```

```python
i = 1
for video in f:
    extract_audio_with_wrapper(video, f'content{i}.mp3')
    i+=1
```

```python
import whisper
import librosa
import noisereduce as nr
import soundfile as sf

# Define the path in Google Drive
gdrive_path = "/content/drive/MyDrive/cleaned_audio/"

# Create the directory if it doesn't exist
import os
os.makedirs(gdrive_path, exist_ok=True)

# Audio file to process


for i in range(1,20):
    data, rate = librosa.load(f'content{i}.mp3', sr=16000)
    reduced_noise_audio = nr.reduce_noise(y=data, sr=rate, stationary=False)
    # Save the cleaned audio to Google Drive
    cleaned_path_gdrive = os.path.join(gdrive_path, f'cleaned_audio{i}.mp3')
    sf.write(cleaned_path_gdrive, reduced_noise_audio, rate)
    print(f"Cleaned audio{i} saved to: {cleaned_path_gdrive}")



# 클린업한 오디오를 gddrive_path에 저장함. 따라서 향후 이 파일들을 쓰고 싶으면 거기에 있는 것들 쓰면 됨.
```
